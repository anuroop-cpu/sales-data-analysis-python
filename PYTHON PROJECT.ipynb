#### RENAMING COLUMN NAMES

```python
# (InvoiceNo, StockCode, Description, Quantity, InvoiceDate, UnitPrice, CustomerID, Country)

pj.columns = pj.columns.str.strip()
d =  {'InvoiceNo' :  'Invoice_Num',
     'StockCode' :   'Stockcodes',
     'Description' : 'Desc',
     'Quantity'  :    'Qty',
     'InvoiceDate' : 'InvoiceID',
     'UnitPrice' :   'Item_Price',
     'CustomerID' :'Client_ID',
     'Country' : 'Nation' }

pj.rename(columns=d, inplace=True)
```

### REMOVE NEGATIVE VALUES FROM QUANTITY COLUMN

```python
pj[pj['Qty'] < 0]
```

### ADDING A NEW COLUMN - AMOUNT SPENT

```python
pj['Amount_Spent'] = pj['Qty'] * pj['Item_Price']
pj
```

```python
### REARRANGING COLUMNS FOR MORE READABILITY
```

```python
pj = pj[['Invoice_Num',
         'InvoiceID',
         'Client_ID',
         'Stockcodes',
         'Desc',
         'Qty',
         'Item_Price',
         'Amount_Spent',
         'Nation']]
pj
```

```python
pj.head()
```

### ACCESSING YEAR VALUE FROM INOVICE_DATA

```python
pj['Year'] = pj['InvoiceID'].dt.year
```

```python
pj[['InvoiceID', 'Year']].head()
```

### CHECK TOP 5 NUMBER OF ORDERS

```python
top_5_orders = (
    pj.groupby('Invoice_Num')['Qty']
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

top_5_orders
```

### HOW MUCH MONEY SPENT BY CUSTOMERS

```python
pj.groupby('Client_ID', as_index=False)['Amount_Spent'].sum()
```

### TOP 5 CUSTOMERS WHO SPENT HIGHEST MONEY

```python
pj.groupby('Client_ID')['Amount_Spent'] \
  .sum() \
  .sort_values(ascending=False) \
  .head(5)
```

### NUMBER OF ORDERS FOR DIFFERENT MONTHS

```python
pj.groupby(pj['InvoiceID'].dt.month_name())['Invoice_Num'].nunique()
```

### PLOT THE GRAPH FOR ABOVE QUERY

```python
import matplotlib.pyplot as plt

monthly_orders = pj.groupby(pj['InvoiceID'].dt.month_name())['Invoice_Num'].nunique()

monthly_orders.plot(kind='bar')
plt.title('Number of Orders by Month')
plt.xlabel('Month')
plt.ylabel('Orders')
plt.show()
```

### HOW MANY ORDERS PER DAY?

```python
pj.groupby(pj['InvoiceID'].dt.date)['Invoice_Num'].nunique()
```

### WHAT IS THE DAY WISE SALE COUNT?

```python
pj.groupby(pj['InvoiceID'].dt.day_name())['Amount_Spent'].sum()
```

### HOW MANY ORDERS FOR EACH COUNTRY?

```python
pj.groupby('Nation')['Invoice_Num'].nunique()
```

### DRAW A HORIZONTAL BAR GRAPH FOR ABOVE DATA

```python
country_orders = pj.groupby('Nation')['Invoice_Num'].nunique()

country_orders.sort_values().plot(kind='barh', figsize=(10,8))
plt.title('Orders by Country')
plt.xlabel('Number of Orders')
plt.ylabel('Country')
plt.show()
```
